```python
from pyspark.sql import functions as F

df_traffic_batch = (
    spark.read
        .format("csv")
        .option("header", "true")
        .option("inferSchema", "true")
        .load(
            "Files/bronze/incremental/batch_03/seo_traffic/seo_traffic_batch_03.csv"
        )
)

df_traffic_batch.printSchema()

print("Traffic Batch 02 rows:", df_traffic_batch.count())
```

```python
df_traffic_batch.select(
    F.min("updated_timestamp").alias("min_updated_timestamp"),
    F.max("updated_timestamp").alias("max_updated_timestamp")
).show(truncate=False)
```

```python
traffic_total = df_traffic_batch.count()

traffic_distinct = (
    df_traffic_batch
    .select("traffic_id")
    .distinct()
    .count()
)

print("Total rows:", traffic_total)
print("Distinct traffic IDs:", traffic_distinct)
print("Duplicate IDs:", traffic_total - traffic_distinct)
```

```python
df_keyword_batch = (
    spark.read
        .format("csv")
        .option("header", "true")
        .option("inferSchema", "true")
        .load(
            "Files/bronze/incremental/batch_03/seo_keyword_performance/seo_keyword_performance_batch_03.csv"
        )
)

df_keyword_batch.printSchema()

print(
    "Keyword Performance Batch 01 rows:",
    df_keyword_batch.count()
)
```

```python
df_keyword_batch.select(
    F.min("updated_timestamp").alias("min_updated_timestamp"),
    F.max("updated_timestamp").alias("max_updated_timestamp")
).show(truncate=False)
```

```python
keyword_total = df_keyword_batch.count()

keyword_distinct = (
    df_keyword_batch
    .select("keyword_performance_id")
    .distinct()
    .count()
)

print("Total rows:", keyword_total)
print("Distinct IDs:", keyword_distinct)
print("Duplicate IDs:", keyword_total - keyword_distinct)
```

```python
df_orders_batch = (
    spark.read
        .format("csv")
        .option("header", "true")
        .option("inferSchema", "true")
        .load(
            "Files/bronze/incremental/batch_03/seo_orders/seo_orders_batch_03.csv"
        )
)

df_orders_batch.printSchema()

print(
    "Orders Batch 01 rows:",
    df_orders_batch.count()
)
```

```python
df_orders_batch.select(
    F.min("updated_timestamp").alias("min_updated_timestamp"),
    F.max("updated_timestamp").alias("max_updated_timestamp")
).show(truncate=False)
```

```python
orders_total = df_orders_batch.count()

orders_distinct = (
    df_orders_batch
    .select("order_id")
    .distinct()
    .count()
)

print("Total rows:", orders_total)
print("Distinct order IDs:", orders_distinct)
print("Duplicate IDs:", orders_total - orders_distinct)
```

```python
summary = [
    (
        "Fact_SEO_Traffic",
        df_traffic_batch.count(),
        df_traffic_batch.agg(
            F.min("updated_timestamp")
        ).first()[0],
        df_traffic_batch.agg(
            F.max("updated_timestamp")
        ).first()[0]
    ),
    (
        "Fact_Keyword_Performance",
        df_keyword_batch.count(),
        df_keyword_batch.agg(
            F.min("updated_timestamp")
        ).first()[0],
        df_keyword_batch.agg(
            F.max("updated_timestamp")
        ).first()[0]
    ),
    (
        "Fact_SEO_Orders",
        df_orders_batch.count(),
        df_orders_batch.agg(
            F.min("updated_timestamp")
        ).first()[0],
        df_orders_batch.agg(
            F.max("updated_timestamp")
        ).first()[0]
    )
]

summary_df = spark.createDataFrame(
    summary,
    ["source_table", "row_count", "min_updated_timestamp", "max_updated_timestamp"]
)

summary_df.show(truncate=False)
```
