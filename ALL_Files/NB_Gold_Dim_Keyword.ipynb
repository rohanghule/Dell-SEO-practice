```python
from pyspark.sql import functions as F

df_silver_keyword = (
    spark.read
        .format("delta")
        .load("abfss://Dell_SEO_Analytics@onelake.dfs.fabric.microsoft.com/LH_SEO_Silver.Lakehouse/Tables/dbo/dim_keyword")
)

print("Silver dim_keyword rows:", df_silver_keyword.count())
print("Silver dim_keyword columns:", len(df_silver_keyword.columns))

display(df_silver_keyword.limit(10))
```

```python
from pyspark.sql.window import Window

window_keyword = Window.orderBy("keyword_id")

df_gold_keyword = (
    df_silver_keyword
    .withColumn(
        "keyword_key",
        F.row_number().over(window_keyword)
    )
)

display(
    df_gold_keyword
    .orderBy("keyword_key")
    .limit(10)
)
```

```python
(
    df_gold_keyword.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("dim_keyword")
)
```

```python
df_gold_keyword_check = spark.read.table("dim_keyword")

print("Gold dim_keyword rows:", df_gold_keyword_check.count())
print("Gold dim_keyword columns:", len(df_gold_keyword_check.columns))

display(
    df_gold_keyword_check
    .orderBy("keyword_key")
    .limit(10)
)
```
