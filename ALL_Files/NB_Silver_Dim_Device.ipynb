```python
from pyspark.sql import Row

device_data = [
    Row(device_key=1, device_type="Desktop"),
    Row(device_key=2, device_type="Mobile"),
    Row(device_key=3, device_type="Tablet")
]

df_device = spark.createDataFrame(device_data)

print("Rows:", df_device.count())
print("Columns:", len(df_device.columns))

display(df_device)
```

```python
(
    df_device.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("dim_device")
)
```

```python
df_silver_device = spark.read.table("dim_device")

print("Silver dim_device rows:", df_silver_device.count())
print("Silver dim_device columns:", len(df_silver_device.columns))

display(df_silver_device)
```
