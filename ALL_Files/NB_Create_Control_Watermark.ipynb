```python
from pyspark.sql import Row

watermark_data = [
    Row(
        source_name="SEO",
        table_name="Fact_SEO_Traffic",
        last_watermark="2026-01-31 23:54:00"
    ),
    Row(
        source_name="SEO",
        table_name="Fact_Keyword_Performance",
        last_watermark="2026-01-31 23:50:00"
    ),
    Row(
        source_name="SEO",
        table_name="Fact_SEO_Orders",
        last_watermark="2026-01-31 22:38:00"
    )
]

df_watermark = spark.createDataFrame(watermark_data)

df_watermark = df_watermark.withColumn(
    "last_watermark",
    df_watermark["last_watermark"].cast("timestamp")
)

df_watermark.show(truncate=False)
```

```python
(
    df_watermark.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("control_watermark")
)
```

```python
spark.read.table("control_watermark").show(truncate=False)
```
