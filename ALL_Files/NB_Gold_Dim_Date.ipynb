```python
from pyspark.sql import functions as F

df_silver_date = (
    spark.read
        .format("delta")
        .load("abfss://Dell_SEO_Analytics@onelake.dfs.fabric.microsoft.com/LH_SEO_Silver.Lakehouse/Tables/dbo/dim_date")
)

print("Silver dim_date rows:", df_silver_date.count())
print("Silver dim_date columns:", len(df_silver_date.columns))

display(df_silver_date.orderBy("date").limit(10))
```

```python
df_gold_date = (
    df_silver_date
    .withColumn(
        "date_key",
        F.date_format("date", "yyyyMMdd").cast("int")
    )
)

display(df_gold_date.orderBy("date").limit(10))
```

```python
(
    df_gold_date.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("dim_date")
)
```

```python
df_gold_date_check = spark.read.table("dim_date")

print("Gold dim_date rows:", df_gold_date_check.count())
print("Gold dim_date columns:", len(df_gold_date_check.columns))

display(df_gold_date_check.orderBy("date").limit(10))
```
