```python
from pyspark.sql import functions as F

df_silver_country = (
    spark.read
        .format("delta")
        .load("abfss://Dell_SEO_Analytics@onelake.dfs.fabric.microsoft.com/LH_SEO_Silver.Lakehouse/Tables/dbo/dim_country")
)

print("Silver dim_country rows:", df_silver_country.count())
print("Silver dim_country columns:", len(df_silver_country.columns))
```

```python
from pyspark.sql.window import Window

window_country = Window.orderBy("country_code")

df_gold_country = (
    df_silver_country
    .withColumn(
        "country_key",
        F.row_number().over(window_country)
    )
)

display(
    df_gold_country
    .orderBy("country_key")
)
```

```python
(
    df_gold_country.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("dim_country")
)
```

```python
df_gold_country_check = spark.read.table("dim_country")

print("Gold dim_country rows:", df_gold_country_check.count())
print("Gold dim_country columns:", len(df_gold_country_check.columns))

display(df_gold_country_check)
```
