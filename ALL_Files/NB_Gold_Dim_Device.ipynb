```python
from pyspark.sql import functions as F

df_silver_device = (
    spark.read
        .format("delta")
        .load("abfss://Dell_SEO_Analytics@onelake.dfs.fabric.microsoft.com/LH_SEO_Silver.Lakehouse/Tables/dbo/dim_device")
)

print("Silver dim_device rows:", df_silver_device.count())
print("Silver dim_device columns:", len(df_silver_device.columns))

display(df_silver_device)
```

```python
df_gold_device = df_silver_device.select(
    "device_key",
    "device_type"
)

display(df_gold_device)
```

```python
(
    df_gold_device.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("dim_device")
)
```

```python
df_gold_device_check = spark.read.table("dim_device")

print("Gold dim_device rows:", df_gold_device_check.count())
print("Gold dim_device columns:", len(df_gold_device_check.columns))

display(df_gold_device_check)
```
